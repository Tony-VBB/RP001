```python
## Read the filr and converted to string variable "text"

path="/content/input.txt"

with open(path, "r") as f:
  text=f.read()
```

```python
## using "unique_char" set found unique characters in the text

unique_char=set()

for i in text:
  unique_char.add(i)

unique_char=sorted(unique_char)
```

```python
## Using dictionaries assigned an index to uniques characters as the model understands numbers than words
## dict cti :- key -> character , value -> number
##dict itc:- key -> number, value -> character

cti,itc={},{}

for i,char in enumerate(unique_char):
  cti[char]=i
  itc[i]=char
```

```python
## converting the text into correspnding integer list then to tensor

import torch

text_intl=[cti[char] for char in text]

text_intt=torch.tensor(text_intl)
```

```python
block_size=5
x,y=[],[]
for i in range(0,len(text_intt)-block_size):
  x.append(text_intt[i:i+block_size])
  y.append(text_intt[i+1:i+block_size+1])

```

```python
## token embeddings layer are produced to make the model identify the each character
## position embeddings layer are produced to make model know the position of the character
## Final embeddings token embeddings + vector embeddings ( makes the model know the position and the character)
## token_emb_layer : gives context of the character
## position_emb : gives context of the position of character
## d+model : dimentions of the embeddings

vocab_size=len(unique_char)
d_model=64

token_emb_layer=torch.nn.Embedding(vocab_size,d_model)

position_emb_layer=torch.nn.Embedding(block_size,d_model)
```

```python
class Linear:

  def __init__(self, in_features,out_features):
    self.w=torch.randn(in_features,out_features)*0.02
    self.b=torch.zeros(out_features)
    self.x=None
    self.dw=None
    self.db=None

  def forward(self, x):
    self.x=x
    return torch.matmul(x,self.w)+self.b

  def backward(self,dy):
    b,s, cin=self.x.shape
    cout=dy.shape[-1]

    x_flat =self.x.reshape(-1,cin)
    dy_flat=dy.reshape(-1,cout)

    self.dw=torch.matmul(x_flat.T,dy_flat)
    self.db=dy_flat.sum(dim=0)

    dx=torch.matmul(dy, self.w.T)
```

```python
class Attention:
    def __init__(self, d_model):
        self.d_model = d_model
        self.q = Linear(d_model, d_model)
        self.k = Linear(d_model, d_model)
        self.v = Linear(d_model, d_model)
        self.cache = {}

    def forward(self, x):
        b, s, d = x.shape

        q = self.q.forward(x)
        k = self.k.forward(x)
        v = self.v.forward(x)

        # Scaled dot-product scores
        scale = 1.0 / (d ** 0.5)
        scores = torch.matmul(q, k.transpose(-2, -1)) * scale  # (b, s, s)

        # Causal mask
        mask = torch.tril(torch.ones(s, s)) == 0
        scores = scores.masked_fill(mask, float('-inf'))

        # Softmax probabilities
        attn_weights = F.softmax(scores, dim=-1)  # (b, s, s)

        # Output
        out = torch.matmul(attn_weights, v)  # (b, s, d)

        # Save cache for backward pass
        self.cache = {
            'x': x, 'q': q, 'k': k, 'v': v,
            'scale': scale, 'mask': mask,
            'attn_weights': attn_weights
        }
        return out

    def backward(self, dy):
        x = self.cache['x']
        q = self.cache['q']
        k = self.cache['k']
        v = self.cache['v']

        scale = self.cache['scale']
        mask = self.cache['mask']
        attn_weights = self.cache['attn_weights']

        b, s, d = x.shape

        # 1. Gradient of Attention output wrt Value and Attention Weights
        dv = torch.matmul(attn_weights.transpose(-2, -1), dy)
        d_attn_weights = torch.matmul(dy, v.transpose(-2, -1))

        # 2. Gradient through Softmax
        sum_term = torch.sum(d_attn_weights * attn_weights, dim=-1, keepdim=True)
        d_scores = attn_weights * (d_attn_weights - sum_term)

        # 3. Gradient through Causal Mask
        d_scores = d_scores.masked_fill(mask, 0.0)

        # 4. Gradient through Scaling
        d_scores = d_scores * scale

        # 5. Gradient through q * k^T matrix multiplication
        dq = torch.matmul(d_scores, k)
        dk = torch.matmul(d_scores.transpose(-2, -1), q)

        # 6. Pass gradients backward through q, k, v manual linear layers
        dx_q = self.q.backward(dq)
        dx_k = self.k.backward(dk)
        dx_v = self.v.backward(dv)

        # Total gradient at input x is the sum of gradients from q, k, and v paths
        dx = dx_q + dx_k + dx_v
        return dx
```

```python
class TransformerBlock:
    def __init__(self, d_model):
        self.attn = Attention(d_model)
        self.ffn1 = Linear(d_model, d_model * 4)
        self.ffn2 = Linear(d_model * 4, d_model)

    def forward(self, x):
        self.x_in = x

        # 1. Attention + Residual Connection
        self.attn_out = self.attn.forward(x)
        self.x_after_attn = x + self.attn_out

        # 2. Feed-Forward + Residual Connection (using ReLU)
        self.h_pre_relu = self.ffn1.forward(self.x_after_attn)
        self.h_post_relu = F.relu(self.h_pre_relu)
        self.ffn_out = self.ffn2.forward(self.h_post_relu)

        out = self.x_after_attn + self.ffn_out
        return out

    def backward(self, dy):
        # --- FFN backward branch ---
        d_ffn_out = dy
        d_h_post = self.ffn2.backward(d_ffn_out)

        # ReLU derivative: 1 if h_pre_relu > 0 else 0
        d_h_pre = d_h_post * (self.h_pre_relu > 0).float()
        d_ffn1_in = self.ffn1.backward(d_h_pre)

        # Merge residual gradient with FFN branch gradient
        d_after_attn = dy + d_ffn1_in

        # --- Attention backward branch ---
        d_attn_out = d_after_attn
        d_attn_in = self.attn.backward(d_attn_out)

        # Final input gradient to the block
        dx = d_after_attn + d_attn_in
        return dx
```

```python
class ReLU:

  def __init__(self):
    self.x = None

  def forward(self, x):
    self.x = x
    return torch.maximum(x, torch.tensor(0.0, device=x.device))

  def backward(self, dy):
    # Derivative: 1 if x > 0 else 0
    return dy * (self.x > 0).float()
```

```python
class Softmax:

  def __init__(self):
    self.out = None

  def forward(self, x):
    # Subtract max for numerical stability
    exps = torch.exp(x - torch.max(x, dim=-1, keepdim=True).values)
    self.out = exps / torch.sum(exps, dim=-1, keepdim=True)
    return self.out

  def backward(self, dy):
    # Vector-Jacobian product for softmax
    sum_term = torch.sum(dy * self.out, dim=-1, keepdim=True)
    return self.out * (dy - sum_term)
```

```python
class CrossEntropyLoss:

  def __init__(self):
    self.probs = None
    self.targets = None

  def forward(self, logits, targets):
    # logits: (B, S, vocab_size), targets: (B, S)
    b, s, v = logits.shape

    # Numerical stability shift
    logits_stable = logits - torch.max(logits, dim=-1, keepdim=True).values
    exps = torch.exp(logits_stable)
    self.probs = exps / torch.sum(exps, dim=-1, keepdim=True)

    # Flatten for NLL calculation
    logits_flat = logits_stable.reshape(-1, v)
    targets_flat = targets.reshape(-1)

    correct_logits = logits_flat[torch.arange(logits_flat.shape[0]), targets_flat]
    loss = -correct_logits + torch.log(torch.sum(torch.exp(logits_flat), dim=-1))

    self.targets = targets_flat
    return loss.mean()

```
